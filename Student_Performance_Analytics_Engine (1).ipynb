Student Performance Analytics Engine

College Python Project Using NumPy and Matplotlib

Project Description:

The Student Performance Analytics Engine is a Python-based project designed to analyse student academic performance. It uses NumPy for numerical calculations and Matplotlib for data visualization.

The project calculates total marks, average marks, percentages, grades, student rankings, and pass/fail status. It also identifies students who need academic support and generates graphical reports.

```python
import numpy as np
import matplotlib.pyplot as plt
import csv

print("All libraries imported successfully!")
```

```python
students = np.array([
    "Aarav", "Priya", "Rahul", "Sneha", "Arjun",
    "Ananya", "Rohan", "Isha", "Karan", "Meera",
    "Aditya", "Kavya", "Vikram", "Diya", "Neha"
])

subjects = np.array([
    "Math", "Science", "English", "Computer", "Social"
])

marks = np.array([
    [85, 90, 78, 95, 88],
    [92, 88, 85, 91, 90],
    [65, 70, 68, 75, 72],
    [78, 82, 80, 85, 79],
    [55, 60, 58, 65, 62],
    [95, 96, 92, 98, 94],
    [45, 50, 48, 55, 52],
    [88, 85, 90, 92, 87],
    [35, 40, 38, 45, 42],
    [80, 83, 85, 88, 82],
    [72, 75, 70, 78, 74],
    [90, 92, 88, 95, 91],
    [60, 65, 62, 68, 64],
    [98, 95, 97, 99, 96],
    [50, 55, 52, 60, 58]
])

print("Student dataset created successfully!")
print("Total students:", len(students))
print("Total subjects:", len(subjects))
print("Marks shape:", marks.shape)
```

```python
if marks.ndim != 2:
    raise ValueError("Marks must be a 2D array.")

if marks.shape[0] != len(students):
    raise ValueError("Student count does not match marks.")

if marks.shape[1] != len(subjects):
    raise ValueError("Subject count does not match marks.")

if np.any((marks < 0) | (marks > 100)):
    raise ValueError("Marks must be between 0 and 100.")

print("Data validation successful!")
```

```python
total_marks = np.sum(marks, axis=1)

print("STUDENT TOTAL MARKS")
print("-" * 40)

for i in range(len(students)):
    print(students[i], ":", total_marks[i])
```

```python

average_marks = np.mean(marks, axis=1)


percentage = (average_marks / 100) * 100


print("STUDENT PERFORMANCE REPORT")
print("-" * 50)

for i in range(len(students)):
    print(f"{students[i]} : Average = {average_marks[i]:.2f}, Percentage = {percentage[i]:.2f}%")
```

```python
# Grade Calculation

grades = []

for avg in average_marks:
    if avg >= 90:
        grades.append("A+")
    elif avg >= 80:
        grades.append("A")
    elif avg >= 70:
        grades.append("B")
    elif avg >= 60:
        grades.append("C")
    elif avg >= 40:
        grades.append("D")
    else:
        grades.append("F")

# Display Grades
print("STUDENT GRADES")
print("-" * 30)

for i in range(len(students)):
    print(f"{students[i]} : {grades[i]}")
```

```python
# Pass/Fail Analysis

results = []

for avg in average_marks:
    if avg >= 40:
        results.append("Pass")
    else:
        results.append("Fail")

# Display Results
print("PASS / FAIL ANALYSIS")
print("-" * 30)

for i in range(len(students)):
    print(f"{students[i]} : {results[i]}")

# Count Pass and Fail Students
pass_count = results.count("Pass")
fail_count = results.count("Fail")

print("\nTotal Passed Students:", pass_count)
print("Total Failed Students:", fail_count)
```

```python
# Topper Analysis

topper_index = np.argmax(average_marks)

print("🏆 CLASS TOPPER")
print("-" * 40)

print("Topper Name:", students[topper_index])
print("Average Marks:", average_marks[topper_index])
print("Percentage:", percentage[topper_index])
print("Grade:", grades[topper_index])
```

```python
# Subject-wise Performance Analysis

subject_average = np.mean(marks, axis=0)

print("SUBJECT-WISE PERFORMANCE")
print("-" * 40)

for i in range(len(subject_average)):
    print(f"Subject {i+1}: Average Marks = {subject_average[i]:.2f}")

# Highest Performing Subject
best_subject = np.argmax(subject_average)

print("\nBest Performing Subject:", best_subject + 1)
print("Highest Average Marks:", subject_average[best_subject])
```

```python

# Data Visualization

import matplotlib.pyplot as plt

# Graph 1: Student Average Marks

plt.figure(figsize=(10, 5))

plt.bar(students, average_marks)

plt.title("Student Average Marks")
plt.xlabel("Students")
plt.ylabel("Average Marks")

plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()


# Graph 2: Subject-wise Average Marks

subject_names = ["Maths", "Science", "English", "Computer", "Hindi"]

plt.figure(figsize=(8, 5))

plt.bar(subject_names[:len(subject_average)], subject_average)

plt.title("Subject-wise Performance")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.grid(axis='y')

plt.show()
```

```python
# Student Performance Dashboard

import pandas as pd

# Create Performance Dashboard

dashboard = pd.DataFrame({
    "Student Name": students,
    "Average Marks": average_marks,
    "Percentage": percentage,
    "Grade": grades,
    "Result": results
})

# Display Dashboard

print("STUDENT PERFORMANCE ANALYTICS DASHBOARD")
print("=" * 60)

display(dashboard)

# Overall Class Statistics

print("\nCLASS STATISTICS")
print("-" * 40)

print("Total Students:", len(students))
print("Class Average:", np.mean(average_marks))
print("Highest Marks:", np.max(average_marks))
print("Lowest Marks:", np.min(average_marks))
print("Total Passed:", pass_count)
print("Total Failed:", fail_count)
```

```python
# Advanced Statistical Analysis

print("ADVANCED STATISTICAL ANALYSIS")
print("=" * 50)

# 1. Mean Marks
print("Mean Marks:", np.mean(average_marks))

# 2. Median Marks
print("Median Marks:", np.median(average_marks))

# 3. Maximum Marks
print("Maximum Average:", np.max(average_marks))

# 4. Minimum Marks
print("Minimum Average:", np.min(average_marks))

# 5. Standard Deviation
print("Standard Deviation:", np.std(average_marks))

# 6. Variance
print("Variance:", np.var(average_marks))

# 7. Student Ranking

ranking = np.argsort(average_marks)[::-1]

print("\nSTUDENT RANKING")
print("-" * 40)

for rank, index in enumerate(ranking, start=1):
    print(f"Rank {rank}: {students[index]} - {average_marks[index]:.2f}")
```
